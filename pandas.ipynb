## Умовний вибір (булеві масиви)

Ми бачили умовний вибір, застосований в `numpy`, і він працюватиме таким же чином для `DataFrame`.

```python
df['Continent']
```

```python
df['Continent'] == "Europe"
```

```python
mask = df['Continent'] == 'Europe'
mask
```

```python
df[mask]
```

```python
df[df['Continent'] == "Europe"]
```

```python
# декілька умов
# європейське з HDI >= 0.9

mask_europe = df['Continent'] == 'Europe'
mask_hdi = df['HDI'] >= 0.9

mask = mask_europe & mask_hdi

df[mask]
```

```python
df[(df['Continent'] == 'Europe') | (df['HDI'] >= 0.9)]
```

```python
df[~(df['Continent'] == 'Europe')]
```

```python
mask = df['Continent'] == 'Europe'  # чи в Європі
mask = ~mask    # не в Європі
df[mask]
```

```python
# & -- and
# | -- or
# ~ -- not
```

```python
df
```

```python
df['Population'] > 70
```

```python
df[df['Population'] > 70]
```

Булева маска дозволяє вибрати потрібні рядки, враховуючи всі стовпчики. Якщо потрібно вибрати лише деякі стовпці, то використовуємо метод `.loc`

```python
df.loc[df['Population'] > 70, ['Name']]
```

```python
df.loc[df['Population'] > 70, ['Name', 'Population', 'GDP']]
```

```python
df.drop(columns=["HDI"])
```

```python
df.drop(['Population', 'HDI'], axis='columns')
```

Усі ці методи `drop` повертають новий `DataFrame`. Якщо ви хочете змінити його "на місці", ви можете скористатися атрибутом `inplace`
